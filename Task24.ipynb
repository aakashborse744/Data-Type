## List iteration with string.

```python
a = ["tushar","rohit","mayur","nilesh","shekhar"]
for i in range(len(a)):
        print(i ,"------->", a[i])
```

```python
for i in range(len(a)-1,-1,-1):
        print(i ,"------->", a[i])
```

```python
## Reverse only index not string.
for i in range(len(a)):
    print(i,"-------->", a[len(a)-1-i]) 
```

```python
## Reverse name 
for i in range(len(a)):
    print(a[i],"-------->",a[i][::-1])
```

```python
## Ends with check
for i in range(len(a)):
    if a[i].endswith('r'):
        print(a[i])
```

```python
## starts with check
for i in range(len(a)):
    if a[i].startswith('n'):
        print(a[i])
```

```python
## filter names longer than 5
for i in range(len(a)):
    if len(a[i]) == 5:
        print(a[i])

## more than 5 character

# for i in range(len(a)):
#     if len(a[i]) > 5:
#         print(a[i])
```

```python
## capitalize first letter
for i in range(len(a)):
    print(a[i].capitalize())
```

```python
## length of each name
for i in range(len(a)):
    print(a[i],"------->",len(a[i]))
```

```python
for i in range(len(a)):
    print(a[i].upper())
```
